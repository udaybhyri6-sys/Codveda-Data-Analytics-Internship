# Level 1- Task 3: Exploratory Data Analysis (EDA)
## Introduction
Exploratory Data Analysis (EDA) is a crucial step in the data science process that helps in understanding the structure, quality, and characteristics of a dataset before applying machine learning algorithms. It involves summarizing the main features of the data using statistical techniques and visualizations to discover hidden patterns, identify relationships between variables, detect anomalies, and gain meaningful insights.

In this task, the cleaned Titanic dataset obtained from the preprocessing stage is analyzed using various statistical measures and visualization techniques. Summary statistics, histograms, box plots, scatter plots, count plots, and correlation analysis are performed to better understand passenger demographics, fare distribution, survival patterns, and relationships among different features.

The insights obtained from this analysis provide a strong foundation for selecting appropriate features and developing accurate machine learning models in the subsequent stages of the project.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use("ggplot")
```

```python
df=pd.read_csv("cleaned_iris.csv")
print("Dataset Loaded Successfully")
df.head()
```

```python
print("Shape of the dataset", df.shape)
print("\ncolumn Names:")
df.columns
print("\nInformation:")
df.info()
```

```python
print("Summary Statistics")
df.describe()
print("\nmedian")
df.median(numeric_only=True)
print("\nvariance")
df.var(numeric_only=True)
```

```python
df.hist(figsize=(12,8))
plt.suptitle("histogram of numerical features")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df)
plt.title("Boxplot of Features")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df,x=df.columns[0],y=df.columns[1],hue=df.columns[-1])
plt.title("Scatter Plot")
plt.show()
```

```python
plt.figure(figsize=(8,6))
corr=df.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
sns.pairplot(df,hue=df.columns[-1])
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x=df.iloc[:,-1])
plt.title("Species Distribution")
plt.show()
```
